```python

import pandas as pd

carsOG = pd.read_csv("Cars_Alt.csv")

carsOG.insert(0, "Observation ID", range(1, len(carsOG) + 1))

yearMapping = {
    "Pre 05": (2004, 2004),
    "05-09": (2005, 2009),
    "10-14": (2010, 2014),
    "15-19": (2015, 2019),
    "20-26": (2020, 2026)
}

carsOG["Year Start"] = carsOG["Year_Range"].map(
    lambda x: year_mapping[x][0]
)
carsOG["Year End"] = carsOG["Year_Range"].map(
    lambda x: year_mapping[x][1]
)
carsOG["Estimated Year Midpoint"] = (
    carsOG["Year Start"] + carsOG["Year End"]
) / 2

carsOG["Is Modified"] = carsOG["Modified"].map({
    "Yes": 1,
    "No": 0
})

# Step 6: Export processed dataset
carsOG.to_csv("Cars_Cleaned.csv", index=False)

print(carsOG.head())
```
