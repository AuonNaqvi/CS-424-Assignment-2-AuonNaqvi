```python
import pandas as pd
import plotly.express as px

evalData = pd.read_csv("C:\\Users\\auon1\\Documents\\CS424Assignment2\\CS-424-Assignment-2-AuonNaqvi\\Assignment3EvalData.csv")

evalData.groupby("Vis")["Time"].describe()

```

```python
evalData.groupby("Vis")["Answer accuracy"].agg(["count", "sum", "mean"])
```

```python
evalData.groupby(["Question", "Vis"])["Time"].describe()
```

```python
evalData.groupby(["ID", "Vis"])["Time"].describe()
```

```python
medianAccuracy = evalData.groupby(["Question", "Vis"], as_index=False)["Answer accuracy"].median()

medAccuracyBarChart = px.bar(medianAccuracy, x="Question", y="Answer accuracy", color="Vis", barmode="group", title="Median Accuracy by Question")
medAccuracyBarChart.show()
```

```python
medianTime = evalData.groupby(["Question", "Vis"], as_index=False)["Time"].median()

medTimeBarChart = px.bar(medianTime, x="Question", y="Time", color="Vis", barmode="group", title="Median Completion Time by Question")
medTimeBarChart.show()
```
